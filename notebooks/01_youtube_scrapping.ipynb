```python
from youtube_transcript_api import YouTubeTranscriptApi

api = YouTubeTranscriptApi()

print("YouTubeTranscriptApi initialized successfully")
```

```python
video_id = "E8zpgNPx8jE"

transcript = api.fetch(video_id)

print(type(transcript))
print(len(transcript))
```

```python
transcript[0]
print(transcript[0].text)
print(transcript[0].start)
print(transcript[0].duration)
```

```python
full_text = " ".join(
    snippet.text for snippet in transcript
)
print(full_text[:2000])
```

```python
empty_count = sum(
    1 for snippet in transcript
    if not snippet.text.strip()
)

print("Total snippets:", len(transcript))
print("Empty snippets:", empty_count)
```

```python
print(transcript.language)
print(transcript.language_code)
print(transcript.is_generated)
```

```python
last = transcript[-1]

print("Last text:", last.text)
print("Last start:", last.start)
print("Last duration:", last.duration)
print("Approx. transcript end:", last.start + last.duration)
```

```python
transcript_data = [
    {
        "text": snippet.text,
        "start": snippet.start,
        "duration": snippet.duration
    }
    for snippet in transcript
]

print(transcript_data[:3])
```

```python
clean_text = " ".join(
    snippet.text.strip()
    for snippet in transcript
    if snippet.text.strip()
)

print(clean_text[:2000])
```

```python
print("Raw characters:", sum(len(s.text) for s in transcript))
print("Clean characters:", len(clean_text))
```
