```python
import requests

channel_url = "https://www.youtube.com/@SkyNews"

response = requests.get(
    channel_url,
    headers={
        "User-Agent": "Mozilla/5.0"
    },
    timeout=20
)

print("Status:", response.status_code)
print("Content length:", len(response.text))
print("Contains videoRenderer:", "videoRenderer" in response.text)
```

```python
import re

video_ids = re.findall(
    r'"videoId":"([A-Za-z0-9_-]{11})"',
    response.text
)

video_ids = list(dict.fromkeys(video_ids))

print("Unique video IDs found:", len(video_ids))
print("First 10 IDs:", video_ids[:10])
```

```python
video_id = video_ids[0]

video_url = f"https://www.youtube.com/watch?v={video_id}"

video_response = requests.get(
    video_url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=20
)

print("Video ID:", video_id)
print("Status:", video_response.status_code)
print("URL:", video_url)
print("Contains title:", '"title"' in video_response.text)
```

```python
import re

match = re.search(
    r'"title":"([^"]+)"',
    video_response.text
)

if match:
    print("Title:", match.group(1))
else:
    print("Title not found")
```

```python
import re

date_match = re.search(
    r'"publishDate":"([^"]+)"',
    video_response.text
)

print("Publish date:", date_match.group(1) if date_match else "Not found")
```

```python
from datetime import datetime, timedelta, timezone

cutoff = datetime.now(timezone.utc) - timedelta(days=2)

recent_video_ids = []

for video_id in video_ids:
    video_url = f"https://www.youtube.com/watch?v={video_id}"

    video_response = requests.get(
        video_url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=10
    )

    if video_response.status_code != 200:
        continue

    date_match = re.search(
        r'"publishDate":"([^"]+)"',
        video_response.text
    )

    if not date_match:
        continue

    published_at = datetime.fromisoformat(date_match.group(1))

    if published_at >= cutoff:
        recent_video_ids.append({
            "video_id": video_id,
            "published_at": published_at,
            "url": video_url
        })

print("Recent videos:", len(recent_video_ids))

for video in recent_video_ids:
    print(video)
```

```python
from datetime import datetime, timedelta, timezone
import requests
import feedparser


def fetch_recent_videos(channel_id: str, lookback_hours: int = 24):
    """
    Fetch recent videos from a YouTube channel using its RSS feed.

    Parameters
    ----------
    channel_id : str
        YouTube channel ID.
    lookback_hours : int
        Number of hours to look back from the current UTC time.

    Returns
    -------
    list[dict]
        Recent video metadata.
    """

    rss_url = (
        f"https://www.youtube.com/feeds/videos.xml"
        f"?channel_id={channel_id}"
    )

    response = requests.get(
        rss_url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=20
    )

    response.raise_for_status()

    feed = feedparser.parse(response.content)

    cutoff = datetime.now(timezone.utc) - timedelta(hours=lookback_hours)

    videos = []

    for entry in feed.entries:
        published_at = datetime.fromisoformat(entry.published)

        # Ignore videos outside the lookback window
        if published_at < cutoff:
            continue

        # Ignore YouTube Shorts
        if "/shorts/" in entry.link:
            continue

        videos.append({
            "video_id": entry.yt_videoid,
            "title": entry.title,
            "published_at": published_at,
            "url": entry.link,
        })

    return videos
```

```python
channel_id = "UCoMdktPbSTixAyNGwb-UYkQ"

recent_videos = fetch_recent_videos(
    channel_id=channel_id,
    lookback_hours=24
)

print("Recent videos:", len(recent_videos))

for video in recent_videos:
    print(
        video["published_at"],
        "|",
        video["video_id"],
        "|",
        video["title"]
    )
```
